### Interpretation

#### Rotation Invariance
The results clearly show that both the classification and segmentation models are not fully rotation invariant. Performance gradually degrades as the rotation angle increases, with a dramatic drop at 90° rotation. This suggests that:

1. The model has learned spatial orientation features that are not invariant to rotation.
2. The absence of a transformation network block (which was excluded per assignment instructions) contributes to this lack of rotation invariance.
3. At 90° rotation, the model encounters point clouds in orientations it has never seen during training, causing significant confusion.

For classification, the model confuses object classes when their canonical orientation is changed. For segmentation, the model struggles to identify parts that have changed their relative spatial positions due to rotation.

#### Point Density Sensitivity
Both models show strong robustness to variations in point density:

1. **Higher Density (10,000 points)**: Both models actually perform slightly better with more points, as additional geometric details help the model make more accurate predictions.

2. **Lower Density (1,000 points)**: Performance decreases but not dramatically for classification, suggesting the model can still recognize global shape features even with fewer points. 

## Q4. Bonus Question - Locality (PointNet++)

For this bonus question, I implemented PointNet++ to incorporate locality information, which is a hierarchical extension of PointNet that processes points in a nested partitioning of the input point set.

### Implementation Details
PointNet++ uses a hierarchical approach with Set Abstraction layers that:
1. Sample a subset of points
2. Group neighboring points around the sampled points
3. Process the grouped points with a mini-PointNet to extract local features

This hierarchical structure allows the network to capture both local geometric structures at varying scales and global shape information.
